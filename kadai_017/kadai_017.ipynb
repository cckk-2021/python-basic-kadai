```python
class Human:
  def __init__(self, name, age):
      self.name = name
      self.age = age

  def check_adult(self):
      if self.age >= 20:
          print(f"{self.name}は大人です。")
      else:
          print(f"{self.name}は子供です。")
humans = [
    Human("taro",30),
    Human("hanako", 19),
    Human("ichiro", 89),
    Human("kikuko", 101)
]

for human in humans:
    human.check_adult()
```
